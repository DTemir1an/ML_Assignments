### Тапсырма 22. Үшбұрыштың түрі

Үшбұрыштың үш қабырғасының ұзындығы `a`, `b`, `c` берілген. Алдымен мұндай үшбұрыштың болуы мүмкін екенін тексеріңіз: кез келген екі қабырғаның қосындысы үшінші қабырғадан үлкен болуы керек. Үшбұрыш бар болса, оның тең қабырғалы, теңбүйірлі немесе әртүрлі қабырғалы екенін анықтаңыз.

````python
# Шешіміңізді осында жазыңыз
a = 10
b = 12
c = 1
isTriangle = None

if a + b > c and a + c > b and b + c > a: isTriangle = True
else: isTriangle = False

if isTriangle and a == b and a == c: typeOfTr = "Equilateral"
elif isTriangle and (a == b or a == c or b == c): typeOfTr = 'Isosceles'
elif isTriangle and not ((a == b or a == c or b == c) and (a == b and a == c)): typeOfTr = "Usual type"
elif not isTriangle: typeOfTr = 'Is not a triangle'

print(typeOfTr)

````

### Тапсырма 23. Қарапайым калькулятор

`a`, `b` сандары және `operation` айнымалысы (`"+"`, `"-"`, `"*"` немесе `"/"`) берілген. Көрсетілген амалды орындап, нәтижені шығарыңыз. Нөлге бөлуге әрекет жасалса немесе белгісіз амал берілсе, тиісті хабарлама шығарыңыз.

````python
# Шешіміңізді осында жазыңыз
a = 2
b = 7
operation = '/'

match operation:
  case '+': print(a+b)
  case '-': print(a-b)
  case '*': print(a*b)
  case '/': print(a/b) if b != 0 else print("Cannot divide by zero")
  case _: print("Unknown operation")

````

### Тапсырма 28. Сандарды санау

`numbers = [12, -5, 8, 0, -3, 17, 4, 0, -9]` тізімі берілген. Тізімдегі оң, теріс сандардың және нөлдердің санын бөлек есептеп шығарыңыз.

**Күтілетін нәтиже:**

```
Оң сандар: 4
Теріс сандар: 3
Нөлдер: 2
```

````python
# Шешіміңізді осында жазыңыз
numbers = [12, -5, 8, 0, -3, 17, 4, 0, -9]
neg = 0
pos = 0
zero = 0

for i in numbers:
  if i == 0: zero += 1
  elif i > 0: pos += 1
  else: neg += 1

print(f"""positive numbers: {pos}
negative numbers: {neg}
zeros: {zero}""")
````

### Тапсырма 44. Тізімнің статистикасы

Сандар тізімін қабылдап, оның ең кіші мәнін, ең үлкен мәнін және орташа мәнін кортеж түрінде қайтаратын `stats(numbers)` функциясын жазыңыз. `min()` және `max()` функцияларын қолданбаңыз. Бос тізім берілсе, функция `None` қайтаруы керек.

**Күтілетін нәтиже:**

```
stats([4, 9, 1, 7, 4]) -> (1, 9, 5.0)
stats([]) -> None
```

````python
# Шешіміңізді осында жазыңыз
def stats(nums: list):
  max = 0
  min = 999999
  sum = 0
  for i in nums:
    max = i if i > max else max
    min = i if i < min else min
    sum += i

  if len(nums) == 0:
    return None
  else:
    avr = sum / len(nums)
    return (min, max, avr)

print(stats([4, 9, 1, 7, 4]))
print(stats([]))


````

```
word_frequency("Python is easy and Python is fun")
-> {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'fun': 1}
```

````python
# Шешіміңізді осында жазыңыз
def word_frequency(text: str):
  result = {}
  splitted = text.split()
  count = 0
  for i in splitted:
    for j in splitted:
      count += 1 if j == i else 0
    result[i] = count
    count = 0

  return result

print(word_frequency("Python is easy and Python is fun"))
````

Құпиясөздің сенімділігін тексеретін `validate_password(password)` функциясын жазыңыз. Құпиясөз кемінде 8 символдан тұруы, кемінде бір цифр және кемінде бір бас әріп қамтуы керек. Функция құпиясөз сенімді болса "Құпиясөз сенімді" деп, әйтпесе қай талап орындалмағанын көрсететін хабарламаны қайтаруы тиіс. Жол әдістерінен `isdigit()` және `isupper()` пайдалы болуы мүмкін.

**Күтілетін нәтиже:**

```
validate_password("abc") -> Құпиясөз кемінде 8 символдан тұруы керек
validate_password("abcdefgh1") -> Құпиясөзде бас әріп болуы керек
validate_password("Abcdefgh1") -> Құпиясөз сенімді
```

````python
# Шешіміңізді осында жазыңыз
def validate_password(password: str):
  hasNum = False
  hasCap = False

  for i in password:
    if i.isdigit(): hasNum = True
    elif i.isupper(): hasCap = True

  if len(password) >= 8 and hasNum and hasCap: return "Good password"
  elif len(password) >= 8 and hasNum and not hasCap: return "Has to have cap letter"
  elif len(password) >= 8 and not hasNum and hasCap: return "Has to have num"
  elif len(password) < 8: return "has to have at least 8 symbols"

print(validate_password("abc"))
print(validate_password("abcdefgh1"))
print(validate_password("Abcdefgh1"))
````

### Тапсырма 47. Банкомат симуляциясы

Бастапқы баланс `balance = 10000`. Операциялар тізімі берілген:

`operations = [("deposit", 5000), ("withdraw", 3000), ("withdraw", 15000), ("deposit", 2000), ("withdraw", 4000)]`

Әр операцияны орындайтын `process(balance, operation)` функциясын жазыңыз. `deposit` балансты көбейтеді, `withdraw` азайтады. Қаражат жеткіліксіз болса, операция орындалмай, ескерту шығуы керек. Барлық операциядан кейінгі соңғы балансты шығарыңыз.

**Күтілетін нәтиже:**

```
Соңғы баланс: 10000
```

````python
# Шешіміңізді осында жазыңыз
balance = 10000
operations = [("deposit", 5000), ("withdraw", 3000), ("withdraw", 15000), ("deposit", 2000), ("withdraw", 4000)]

def process(balance, operation):
    action, amount = operation

    if action == "deposit": balance += amount
    elif action == "withdraw":
        if balance >= amount: balance -= amount
        else: print("not enough funds")

    return balance

for operation in operations:
  balance = process(balance, operation)

print(f"final balance: {balance}")

````

### Тапсырма 48. Студенттер журналы

Студенттердің баллдары сөздікте берілген:

`journal = {"Aigerim": [85, 92, 78], "Nurlan": [65, 70, 58], "Dana": [95, 98, 91], "Arman": [45, 52, 40]}`

Келесі функцияларды жазыңыз: орташа баллды есептейтін `average(scores)`, баллға сәйкес бағаны (A, B, C, F) қайтаратын `get_grade(score)`, үздік студенттің атын қайтаратын `best_student(journal)`. Осы функцияларды қолданып, әр студенттің аты, орташа баллы мен бағасы көрсетілген кестені және үздік студентті шығарыңыз.

````python
# Шешіміңізді осында жазыңыз
journal = {"Aigerim": [85, 92, 78], "Nurlan": [65, 70, 58], "Dana": [95, 98, 91], "Arman": [45, 52, 40]}

def average(scores: list):
    theSum = 0
    for i in scores:
        theSum += i

    avrScore = theSum / len(scores)

    return round(avrScore, 2)


def get_grade(score: float):
    if 90 <= score <= 100:
        return "A"
    elif 75 <= score < 90:
        return "B"
    elif 60 <= score < 75:
        return "C"
    elif 0 <= score < 60:
        return "F"
    else:
        return "invalid score"


def best_student(journal: dict):
    avrScores = {}

    for student, scores in journal.items():
        avrScores[student] = average(scores)

    maxScore = max(avrScores.values())

    for student, score in avrScores.items():
        if score == maxScore:
            return student


for student, scores in journal.items():
    avg = average(scores)
    grade = get_grade(avg)

    print(f"{student}: average = {avg}, grade = {grade}")

print("best student:", best_student(journal))
````

````python
# Шешіміңізді осында жазыңыз
def encrypt(text: str, shift: int):
  encrypted = []
  for i in text:
    if (65 <= ord(i) <= (90 - shift)) or (97 <= ord(i) <= (122 - shift)):
      ascVal = ord(i) + shift
      encrypted.append(chr(ascVal))
    elif (90 >= ord(i) > (90 - shift)):
      ascVal = ord(i) + shift - 90 + 64
      encrypted.append(chr(ascVal))
    elif 122 >= ord(i) > (122 - shift):
      ascVal = ord(i) + shift - 122 + 96
      encrypted.append(chr(ascVal))
    else: encrypted.append(i)

  return encrypted



def decrypt(text: str, shift: int):
  decrypted = []
  for i in text:
    if ((65 + shift) <= ord(i) <= 90) or ((97 + shift) <= ord(i) <= 122):
      ascVal = ord(i) - shift
      decrypted.append(chr(ascVal))
    elif (65 <= ord(i) < (65 + shift)):
      ascVal = 90 - (64 + shift - ord(i))
      decrypted.append(chr(ascVal))
    elif 97 <= ord(i) < (97 + shift):
      ascVal = 122 - (96 + shift - ord(i))
      decrypted.append(chr(ascVal))
    else: decrypted.append(i)
  return decrypted

for i in encrypt("Hello, World!", 3):
  print(i, end = "")

print("\n")
for i in decrypt(encrypt("Hello, World!", 3), 3):
  print(i, end = "")

print("\n")
for i in encrypt("ABC WXYZ, wxyz abc!", 4):
  print(i, end = "")

print("\n")
for i in decrypt(encrypt("ABC WXYZ, wxyz abc!", 4), 4):
  print(i, end = "")

````
